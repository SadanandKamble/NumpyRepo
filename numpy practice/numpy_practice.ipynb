```python
#Homwork :
##create 2 array- 1d and 3D, and find its attributes

import numpy as np

# 1D array
arr_1d = np.array([1, 2, 3, 4, 5])

# 3D array
arr_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

# Attributes of 1D array
print("1D Array:", arr_1d)
print("ndim:", arr_1d.ndim)
print("shape:", arr_1d.shape)
print("size:", arr_1d.size)
print("dtype:", arr_1d.dtype)

print("\n-------------------\n")

# Attributes of 3D array
print("3D Array:", arr_3d)
print("ndim:", arr_3d.ndim)
print("shape:", arr_3d.shape)
print("size:", arr_3d.size)
print("dtype:", arr_3d.dtype)
```
