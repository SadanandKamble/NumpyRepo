```python
"""Employee Salary & Performance Analysis

A company has 5 employees and stores their monthly salary and performance scores in 4 quarters. They want to:

1.Store the data in a NumPy array.
2.Calculate total annual salary for each employee.
3.Find the average performance score for each employee.
4.Determine the highest salary in each quarter.
5.Flatten and reshape the data for analytics.
6.Slice the data to get specific employee info.
"""
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Employee Data (rows: employees, columns: [Q1 Salary, Q2 Salary, Q3 Salary, Q4 Salary])
# Also include Performance Scores for each quarter
salary = np.array([
    [50000, 52000, 51000, 53000],  # Employee 1
    [60000, 61000, 60500, 62000],  # Employee 2
    [55000, 54000, 56000, 57000],  # Employee 3
    [48000, 49000, 50000, 49500],  # Employee 4
    [70000, 71000, 72000, 73000]   # Employee 5
])

performance = np.array([
    [8, 9, 7, 9],   # Employee 1
    [7, 8, 8, 7],   # Employee 2
    [9, 9, 8, 9],   # Employee 3
    [6, 7, 6, 7],   # Employee 4
    [10, 9, 10, 10] # Employee 5
])

print("Salary Data:\n", salary)
print("\nPerformance Data:\n", performance)
```

```python
# 2: Total annual salary per employee
total_salary = salary.sum(axis=1)
print("\nTotal Annual Salary per Employee:", total_salary)

# 3: Average performance per employee
avg_performance = performance.mean(axis=1)
print("Average Performance per Employee:", avg_performance)

# 4: Highest salary in each quarter
highest_salary_per_quarter = salary.max(axis=0)
print("Highest Salary in each Quarter:", highest_salary_per_quarter)

# 5: Flatten and reshape salary array
flattened_salary = salary.flatten()
reshaped_salary = salary.reshape(4, 5)  # 4 quarters, 5 employees
print("\nFlattened Salary Array:", flattened_salary)
print("Reshaped Salary Array (4x5):\n", reshaped_salary)

# 6: Slice example: Salary of first 3 employees in first 2 quarters
slice_salary = salary[:3, :2]
print("\nSalary of first 3 employees in first 2 quarters:\n", slice_salary)




```
