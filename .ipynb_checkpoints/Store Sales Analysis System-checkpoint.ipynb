```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Create sales data for 5 products over 4 weeks (explicit values)
sales = np.array([
    [120, 150, 130, 140],  # Product 1
    [200, 180, 160, 190],  # Product 2
    [90,  110, 105, 115],  # Product 3
    [300, 320, 310, 330],  # Product 4
    [50,  60,  55,  65]    # Product 5
])

products = ['Smartphone','Laptop','Smartwatch', 'Wireless Mouse','Bluetooth Speaker']
weeks = ['Week1', 'Week2', 'Week3', 'Week4']

# Step 2: Convert to Pandas DataFrame
df = pd.DataFrame(sales, index=products, columns=weeks)
print("Sales Data:\n", df)

# Step 3: Total and average sales per product
df['Total_Sales'] = df.sum(axis=1)
df['Average_Sales'] = df.mean(axis=1)
print("\nSales with Total and Average:\n", df)

# Step 4: Highest selling product per week
highest_per_week = df[weeks].idxmax()
print("\nHighest Selling Product per Week:\n", highest_per_week)

# Step 5: Flatten and reshape using NumPy
flattened = sales.flatten()
reshaped = sales.reshape(4, 5)  # 4 weeks x 5 products
print("\nFlattened Array:\n", flattened)
print("\nReshaped Array (4x5):\n", reshaped)

# Step 6: Slice example: first 3 products, first 2 weeks
slice_data = sales[:3, :2]
print("\nSlice (First 3 Products, First 2 Weeks):\n", slice_data)
```

```python
#  Bar chart: Total sales per product
plt.figure(figsize=(9,4))
plt.bar(df.index, df['Total_Sales'], color='skyblue')
plt.title('Total Sales per Product')
plt.ylabel('Total Sales')
plt.xlabel('Products')
plt.show()
```

```python
#  Pie chart: Contribution to total sales by each product
plt.figure(figsize=(9,6))
plt.pie(df['Total_Sales'], labels=df.index, autopct='%1.1f%%', startangle=140, colors=plt.cm.Paired.colors)
plt.title('Sales Contribution by Product')
plt.show()
```

```python
# Scatter plot: Weekly sales for each product
plt.figure(figsize=(5,5))
for i, product in enumerate(df.index):
    plt.scatter(weeks, df.loc[product, weeks], s=100, label=product)  # s= marker size
plt.title("Weekly Sales Scatter Plot")
plt.xlabel("Weeks")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt




#  Line chart: Weekly sales trend for each product
plt.figure(figsize=(7,3))
for product in df.index:
    plt.plot(weeks, df.loc[product, weeks], marker='o', label=product)
plt.title('Weekly Sales Trend')
plt.xlabel('Weeks')
plt.ylabel('Sales')
plt.legend()
plt.show()


```
